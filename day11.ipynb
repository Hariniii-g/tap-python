```python
with open(file_name, "r") as f:
    r=csv.DictReader(f)
    oid=cid=ostatus=odate=odate_estimated=odate_delivered=odate_approved=odate_shipped=0
    for i in r:
        if i['order_id'] == "null" or i['order_id'].strip() == "":
            oid+=1
        if i['customer_id'] == "null" or i['customer_id'].strip() == "":
            cid+=1
        if i['order_status'] == "null" or i['order_status'].strip() == "":
            ostatus+=1
        if i['order_purchase_timestamp'] == "null" or i['order_purchase_timestamp'].strip() == "":
            odate+=1  
        if i['order_estimated_delivery_date'] == "null" or i['order_estimated_delivery_date'].strip() == "":
            odate_estimated+=1
        if i['order_delivered_customer_date'] == "null" or i['order_delivered_customer_date'].strip() == "":
            odate_delivered+=1
        if i['order_approved_at'] == "null" or i['order_approved_at'].strip() == "":
            odate_approved+=1  
        if i['order_delivered_carrier_date'] == "null" or i['order_delivered_carrier_date'].strip() == "":
            odate_shipped+=1
       
    print("order id - ",oid,"\ncustomer id - ",cid,"\norder status - ",ostatus,"\norder purchase timestamp - ",odate,"\norder estimated delivery date - ",odate_estimated,"\norder delivered customer date - ",odate_delivered,"\norder approved at - ",odate_approved,"\norder delivered carrier date - ",odate_shipped)
    
```
