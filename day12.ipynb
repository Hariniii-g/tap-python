```python
#Create a column-mapping dictionary to a standard schema.

# Read the CSV into a list of dictionaries
import csv

def read_csv(path):
    with open(path, "r", newline="", encoding="utf-8") as file:
        return list(csv.DictReader(file))


# Convert the column names to the standard names you choose
def apply_column_mapping(rows, column_mapping):
    mapped_rows = []
    for row in rows:
        mapped_row = {}
        for old_name, value in row.items():
            new_name = column_mapping.get(old_name, old_name)
            mapped_row[new_name] = value

        mapped_rows.append(mapped_row)

    return mapped_rows


# Read the Olist orders data
orders = read_csv("olist_mini/orders.csv")

# Map CSV column names to your chosen standard schema
column_mapping = {
    "order_id": "order_id",
    "customer_id": "customer_id",
    "order_status": "status",
    "order_purchase_timestamp": "purchase_time",
    "order_approved_at": "approved_time",
    "order_delivered_carrier_date": "carrier_delivery_time",
    "order_delivered_customer_date": "customer_delivery_time",
    "order_estimated_delivery_date": "estimated_delivery_time",
}

# Apply the mapping
standard_orders = apply_column_mapping(orders, column_mapping)

# Show the schema and one transformed row
print("Standard column names:", list(standard_orders[0].keys()))
print("First row:", standard_orders[0])

    
```

```python
#6. Convert price and freight to float, and timestamps to datetime.

import csv
from datetime import datetime

with open("olist_mini/order_items.csv","r") as f:
    e=0;r=list(csv.DictReader(f)   )
    for i in r:
        e+=1
        print(type(i["price"]))
        print(type(i["freight_value"]))
        print(type(i["shipping_limit_date"]))
        i["price"]=float(i["price"])
        i["freight_value"]=float(i["freight_value"])
        i["shipping_limit_date"]=datetime.strptime(i["shipping_limit_date"],"%Y-%m-%d %H:%M:%S")
        if e>1:
            break
    print("-----------------------------------------------------")
    e=0
    for i in r:
        e+=1
        print(type(i["price"]))
        print(type(i["freight_value"]))
        print(type(i["shipping_limit_date"]))
        if e>1:
            break

    print("the lines are reduced and datatypes converted to float and datetime")

```

```python
# 7. Calculate order totals per order_id using a dictionary.
ot={}
with open("olist_mini/order_items.csv", "r") as f:
    r=csv.DictReader(f)
    for i in r:
        oid= i["order_id"]
        price = float(i["price"])
        fv= float(i["freight_value"])
        total = price + fv

        if oid in ot:
            ot[oid] += total
        else:
            ot[oid] = total
    for i in ot:
        print(i, f"{ot[i]:.2f}")
```
